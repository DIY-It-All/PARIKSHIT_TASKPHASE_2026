```python
#4
import functools

def add_console_logs(func):
  @functools.wraps(func)
  def wrapper(*args,**kwargs):
    print(f"Function {func.__name__} is called")
    result = func(*args,**kwargs)
    print(f"Function {func.__name__} has finished execution")
    return result
  return wrapper

def capitalize(func):
  @functools.wraps(func)
  def wrapper(*args,**kwargs):
    result = func(*args,**kwargs)
    try:
      return result.upper()
    except AttributeError:
      return result
    except Exception as e:
      print(f"An error occurred: {e}")
      return result
  return wrapper

def memoizer(func):
  @functools.wraps(func)
  def wrapper(*args,**kwargs):
    key = (args, tuple(sorted(kwargs.items())))
    if key in wrapper.__cache__:
      print(f"Using cached result for {func.__name__}")
      return wrapper.__cache__[key]
    result = func(*args,**kwargs)
    wrapper.__cache__[key] = result
    return result
  wrapper.__cache__ = {}
  return wrapper

def metadata(**data):
  def decorator(func):
    @functools.wraps(func)
    def wrapper(*args,**kwargs):
      return func(*args,**kwargs)
    for key, value in data.items():
      setattr(wrapper, key, value)
    return wrapper
  return decorator

@add_console_logs
@capitalize
def sample_function():
  return "this is a sample function."

@capitalize
def sample_function_2(name):
  return f"Hello, {name}!"

@memoizer
def multiply(a, b):
  print("Calculating multiplication...")
  return a * b

@metadata(author="Parikshit", category="decorator demo")
def documented_function():
  return "metadata works"

# Testing logging and stacked decorators
print(sample_function())
print(sample_function_2("Alice"))

# Testing memoization: the calculation should print only on the first call
print(multiply(4, 5))
print(multiply(4, 5))

# Testing custom metadata
print(documented_function.author)
print(documented_function.category)
```

```python
#4.5
#generators
def range_custom(start=0,end=10,step=1):
  i = start
  while i < end:
    yield i
    i += step

for i in range_custom(1, 10, 1):
    print(i)
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name} and I am {self.age} years old.")

class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)
        self.student_id = student_id

    def greet(self):
        super().greet()
        print(f"My student ID is {self.student_id}.")

class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.__subject = subject

    def greet(self):
        super().greet()
        print(f"I teach {self.__subject}.")
    def explain_subject(self):
        print(f"My subject is {self.__subject}.")

person = Person("Alice", 30)
person.greet()
student = Student("Bob", 20, "S12345")
student.greet()
teacher = Teacher("Charlie", 40, "Mathematics")
teacher.greet()
teacher.explain_subject()
#teacher.__subject = "Physics" # this will not work as __subject is private
```
