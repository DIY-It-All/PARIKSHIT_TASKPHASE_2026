```python
#numpy

import numpy as np
#import numpy

a = np.array([1, 2, 3])
b = np.array([[1,2,3],[4, 5, 6]])
print(a,a.shape)
print(b,b.shape)
a[2] = 10
print(a)
print(a.ndim, b.ndim)
print(a.size,b.size)
print(a.dtype,b.dtype)

c = np.array([[1,2,3],[4,5,6]], dtype=np.float64)
print(c,c.dtype)

d = np.zeros((3,4))
print(d)
e = np.ones((2,2,2),dtype=np.int16)
print(e)

f = np.empty((1,2))
print(f) # garbage values

g = np.arange(4)
print(g)

g = np.arange(4,9,2)
print(g)


h = np.linspace(0, 10, 5) # np.linspace(0, 10, num=5) is the same
print(h)

i = np.random.random((2,2))
print(i)
```

```python
a = np.array([[1, 4], [3, 1]])
print(a,end="\n\n")
print(np.sort(a, axis=1), end="\n\n") # sort along the first axis


a1 = np.array([[1, 4], [3, 1]])
a2 = np.array([[1, 2], [3, 4]])
print(np.concatenate((a1, a2), axis=0)) # concatenate along the first axis
print(np.concatenate((a1, a2), axis=1)) # concatenate along the second axis
print()

print(np.vstack((a1, a2))) # vertical stack
print(np.hstack((a1, a2))) # horizontal stack

c = np.arange(100)
c = c.reshape(10,10)
print(c,end="\n\n")


d = np.array([1,2,3,4,5,6])
d = d[np.newaxis, :]
print(d, d.shape,end="\n\n")

d = np.array([1,2,3,4,5,6])
d = d[:, np.newaxis]
print(d, d.shape,end="\n\n")
```

```python
a = np.arange(10)
print(a[2:5]) 
print(a[2:])  
print(a[:5])  
print(a[-1])  

print(a[2:5:2])

print(a[::-1])

print(a[a%2==0])

print(a[(a>2) & (a<8)])
x = lambda l : l%5>=4
print(a[x(a)])

```

```python
a = np.arange(12).reshape(3,4)
print(a)
print(a.sum())
print(a.sum(axis=0))
print(a.sum(axis=1))
print(a.mean())
print(a.mean(axis=0))
print(a.prod())
print(a.prod(axis=0))
print(a.transpose())
print(a.T)
print(a.flatten())
print(a.ravel())
print(a.max())
print(a.max(axis=0))
print(a.min())
print(a.min(axis=0))
print(a.argmax())
print(a.argmin())

b = np.array([[1, 2,2,3], [3, 4,3,2]])
print(np.unique(b))
```

```python
a = np.array([[1,2,3],[4,5,6], [9,8,3]])
b = np.array([[7,8,9],[10,11,12], [13,14,15]])
print(np.dot(a,b))
print(a @ b) # same as np.dot(a,b)
print(np.matmul(a,b)) # same as np.dot(a,b)
print(np.inner(a,b)) # inner product of a and b
print(np.outer(a,b)) # outer product of a and b
print(np.cross(a,b)) # cross product of a and b
print(np.linalg.det(a)) # determinant of a
print(np.linalg.inv(a)) # inverse of a
print(np.linalg.eig(a)) # eigenvalues and eigenvectors of a
print(np.linalg.svd(a)) # singular value decomposition of a
print(np.linalg.norm(a)) # Frobenius norm of a
print(np.linalg.matrix_rank(a)) # rank of a
print(np.linalg.solve(a,b)) # solve the linear equation ax = b
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
print(a*1.6)
```
